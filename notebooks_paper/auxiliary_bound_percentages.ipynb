# CPS-2F auxiliary locations as a percentage of their bounds
## Question
How many added matching locations were recorded relative to each curve's own finite construction bound, at alpha=0.5?
This notebook displays saved figures and tables only. It does not run optimization, auxiliary construction, or external validation. Existing evidence and publications are not modified.

```python
from pathlib import Path
import pandas as pd
from IPython.display import Image, Markdown, display
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'output/paper_v1/plots_auxiliary_bound_percentages').exists())
PLOT_DIR = ROOT / 'output/paper_v1/plots_auxiliary_bound_percentages'
DISPLAY_WIDTH = 1100  # Display size only; exported PNGs are 300 dpi.
values = pd.read_csv(PLOT_DIR / 'combined_all45_values.csv')
aliases = pd.read_csv(PLOT_DIR / 'pair_aliases.csv')
assert len(values) == 45 and values.domain.value_counts().to_dict() == {'hurricane': 33, 'protein': 12}
display(values.groupby('domain').agg(pairs=('pair','size'), completed_construction_A=('construction_completed_A','sum'), completed_construction_B=('construction_completed_B','sum')))
```

## Definition and provenance
For curve X, plot $100a_X/[2n_X(n_X-1)]$. Here $a_X$ excludes original vertices; auxiliary locations are matching positions, not selectable output vertices. The finite bound counts at most two sphere–segment intersections for each of n vertex-centered spheres and n−1 segments.
Proteins use the existing **w=16** subsampling. Hurricanes use **all supplied track vertices**, without additional subsampling. Each curve uses its own denominator: normalized percentages neither directly compare absolute counts nor isolate a domain effect. CPS-3F has no auxiliary locations and is omitted.
The CSVs retain integer counts and bounds, exact rational percentage strings, full-precision percentages, attempt paths, JSON fields and source hashes. Bar labels round to one decimal place. The common set requires validated results for all four methods. Stable aliases retain historical numbering and intentional gaps.

```python
limited = values[values.subsequent_search_resource_limited]
display(limited[['alias','pair','nA','nB','aA','aB','bound_A','bound_B','percentage_A','percentage_B','construction_completed_A','construction_completed_B']])
assert len(limited) == 2 and limited[['construction_completed_A','construction_completed_B']].all().all()
display(Markdown('**Asterisks:** auxiliary and single-curve construction completed in both curves; only the subsequent CPS search reached its transition limit. These counts are not incomplete.'))
```
